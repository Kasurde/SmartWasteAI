```python
#Create One Clean Dataset
import os

classes = [
    "battery",
    "biological",
    "brown-glass",
    "cardboard",
    "clothes",
    "green-glass",
    "metal",
    "paper",
    "plastic",
    "shoes",
    "trash",
    "white-glass"
]

base_dir = "/content/waste_dataset"

for cls in classes:
    os.makedirs(os.path.join(base_dir, cls), exist_ok=True)

print("Folders created.")
```

```python
#Remove Duplicate Images
import os
import hashlib
import shutil

source_dir = "/content/all_datasets"
clean_dir = "/content/waste_dataset"

seen = set()
duplicates = 0
copied = 0

valid_extensions = (".jpg", ".jpeg", ".png", ".webp")

for root, dirs, files in os.walk(source_dir):

    for file in files:

        if not file.lower().endswith(valid_extensions):
            continue

        file_path = os.path.join(root, file)

        try:
            with open(file_path, "rb") as f:
                file_hash = hashlib.md5(f.read()).hexdigest()

            if file_hash in seen:
                duplicates += 1
                continue

            seen.add(file_hash)

        except:
            continue

        # Find class from folder name
        class_name = os.path.basename(root).lower()

        if class_name not in classes:
            continue

        destination = os.path.join(
            clean_dir,
            class_name,
            file
        )

        # Avoid filename collision
        if os.path.exists(destination):
            name, ext = os.path.splitext(file)
            destination = os.path.join(
                clean_dir,
                class_name,
                f"{name}_{copied}{ext}"
            )

        shutil.copy2(file_path, destination)
        copied += 1

print("Images copied:", copied)
print("Duplicates removed:", duplicates)
```

Image Preprocessing
Use:
Image Size = 224 × 224

```python
#Image Preprocessing
from tensorflow.keras.preprocessing.image import ImageDataGenerator

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True
)

val_datagen = ImageDataGenerator(
    rescale=1./255
)

test_datagen = ImageDataGenerator(
    rescale=1./255
)
```

```python
#Create Data Generators
train_generator = train_datagen.flow_from_directory(
    "/content/split_dataset/train",
    target_size=(224,224),
    batch_size=32,
    class_mode="categorical"
)

val_generator = val_datagen.flow_from_directory(
    "/content/split_dataset/val",
    target_size=(224,224),
    batch_size=32,
    class_mode="categorical"
)

test_generator = test_datagen.flow_from_directory(
    "/content/split_dataset/test",
    target_size=(224,224),
    batch_size=32,
    class_mode="categorical",
    shuffle=False
)
```

Model 1: CNN
Your first model should be your custom CNN.
Architecture:
Input Image
->
224 × 224 × 3
->
Conv2D
->
MaxPooling
->
Conv2D
->
MaxPooling
->
Conv2D
->
MaxPooling
->
Flatten
->
Dense
->
Dropout
->
12 Classes

```python
import tensorflow as tf
from tensorflow.keras import layers, models

cnn = models.Sequential([

    layers.Conv2D(32, (3,3), activation="relu",
                  input_shape=(224,224,3)),
    layers.MaxPooling2D(),

    layers.Conv2D(64, (3,3), activation="relu"),
    layers.MaxPooling2D(),

    layers.Conv2D(128, (3,3), activation="relu"),
    layers.MaxPooling2D(),

    layers.Flatten(),

    layers.Dense(256, activation="relu"),
    layers.Dropout(0.5),

    layers.Dense(12, activation="softmax")
])

cnn.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

cnn.summary()
```

```python
#Train CNN
history_cnn = cnn.fit(
    train_generator,
    validation_data=val_generator,
    epochs=20
)
```

```python
import os
import shutil
import hashlib
from PIL import Image

# ============================================================
# PATHS
# ============================================================

SOURCE = "/content/extracted_datasets"
OUTPUT = "/content/Waste_Dataset_Clean"

# ============================================================
# FINAL 12 CLASSES
# ============================================================

classes = [
    "battery",
    "biological",
    "brown-glass",
    "cardboard",
    "clothes",
    "green-glass",
    "metal",
    "paper",
    "plastic",
    "shoes",
    "trash",
    "white-glass"
]

# Create output folders
for cls in classes:
    os.makedirs(os.path.join(OUTPUT, cls), exist_ok=True)

# ============================================================
# FOLDER MAPPING
# ============================================================

class_mapping = {
    "battery": "battery",

    "biological": "biological",
    "Food Organics": "biological",
    "Vegetation": "biological",

    "brown-glass": "brown-glass",

    "cardboard": "cardboard",
    "Cardboard": "cardboard",

    "clothes": "clothes",
    "Textile Trash": "clothes",

    "green-glass": "green-glass",

    "metal": "metal",
    "Metal": "metal",

    "paper": "paper",
    "Paper": "paper",

    "plastic": "plastic",
    "Plastic": "plastic",

    "shoes": "shoes",

    "trash": "trash",
    "Miscellaneous Trash": "trash",

    "white-glass": "white-glass"
}

# ============================================================
# IMAGE EXTENSIONS
# ============================================================

image_extensions = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}

# ============================================================
# DUPLICATE TRACKING
# ============================================================

hashes = set()

total_copied = 0
total_duplicates = 0
total_corrupt = 0
total_ignored = 0

# ============================================================
# PROCESS DATASETS
# ============================================================

for dataset in sorted(os.listdir(SOURCE)):

    dataset_path = os.path.join(SOURCE, dataset)

    if not os.path.isdir(dataset_path):
        continue

    print("\nProcessing:", dataset)

    # Walk recursively because some datasets have nested folders
    for root, dirs, files in os.walk(dataset_path):

        # Ignore Mac metadata
        dirs[:] = [
            d for d in dirs
            if d != "__MACOSX"
        ]

        # Check whether current folder is a recognized class
        folder_name = os.path.basename(root)

        if folder_name not in class_mapping:
            continue

        target_class = class_mapping[folder_name]

        for file in files:

            ext = os.path.splitext(file)[1].lower()

            if ext not in image_extensions:
                continue

            source_file = os.path.join(root, file)

            # ------------------------------------------------
            # Check image validity
            # ------------------------------------------------

            try:
                with Image.open(source_file) as img:
                    img.verify()

            except Exception:
                total_corrupt += 1
                continue

            # ------------------------------------------------
            # Calculate SHA-256
            # ------------------------------------------------

            try:
                sha256 = hashlib.sha256()

                with open(source_file, "rb") as f:
                    for chunk in iter(lambda: f.read(1024 * 1024), b""):
                        sha256.update(chunk)

                file_hash = sha256.hexdigest()

            except Exception:
                total_corrupt += 1
                continue

            # ------------------------------------------------
            # Remove exact duplicates
            # ------------------------------------------------

            if file_hash in hashes:
                total_duplicates += 1
                continue

            hashes.add(file_hash)

            # ------------------------------------------------
            # Create unique filename
            # ------------------------------------------------

            new_name = (
                f"{dataset}_"
                f"{total_copied:08d}_"
                f"{file}"
            )

            destination = os.path.join(
                OUTPUT,
                target_class,
                new_name
            )

            shutil.copy2(source_file, destination)

            total_copied += 1


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("CLEANING COMPLETE")
print("=" * 60)

print("Images copied       :", total_copied)
print("Duplicates removed  :", total_duplicates)
print("Corrupt files       :", total_corrupt)
print("Output folder       :", OUTPUT)
```

```python
import os
import shutil
from sklearn.model_selection import train_test_split

# ============================================================
# PATHS
# ============================================================

SOURCE = "/content/Waste_Dataset_Clean"
OUTPUT = "/content/Waste_Dataset_Split"

# ============================================================
# CLASSES
# ============================================================

classes = [
    "battery",
    "biological",
    "brown-glass",
    "cardboard",
    "clothes",
    "green-glass",
    "metal",
    "paper",
    "plastic",
    "shoes",
    "trash",
    "white-glass"
]

# ============================================================
# CREATE TRAIN / VALIDATION / TEST FOLDERS
# ============================================================

for split in ["train", "validation", "test"]:
    for cls in classes:
        os.makedirs(
            os.path.join(OUTPUT, split, cls),
            exist_ok=True
        )

# ============================================================
# SPLIT DATA
# ============================================================

for cls in classes:

    class_folder = os.path.join(SOURCE, cls)

    images = [
        f for f in os.listdir(class_folder)
        if f.lower().endswith(
            (".jpg", ".jpeg", ".png", ".bmp", ".webp")
        )
    ]

    # First: 70% train, 30% temporary
    train_images, temp_images = train_test_split(
        images,
        test_size=0.30,
        random_state=42
    )

    # Second: temporary -> 15% validation + 15% test
    val_images, test_images = train_test_split(
        temp_images,
        test_size=0.50,
        random_state=42
    )

    print(
        f"{cls:15s} "
        f"Train: {len(train_images):5d} | "
        f"Val: {len(val_images):5d} | "
        f"Test: {len(test_images):5d}"
    )

    # ========================================================
    # COPY FILES
    # ========================================================

    for filename in train_images:

        src = os.path.join(class_folder, filename)
        dst = os.path.join(
            OUTPUT, "train", cls, filename
        )

        shutil.copy2(src, dst)

    for filename in val_images:

        src = os.path.join(class_folder, filename)
        dst = os.path.join(
            OUTPUT, "validation", cls, filename
        )

        shutil.copy2(src, dst)

    for filename in test_images:

        src = os.path.join(class_folder, filename)
        dst = os.path.join(
            OUTPUT, "test", cls, filename
        )

        shutil.copy2(src, dst)


print("\n" + "=" * 60)
print("✅ DATASET SPLIT COMPLETE")
print("=" * 60)
print("Train      : 70%")
print("Validation : 15%")
print("Test       : 15%")
print("Location   :", OUTPUT)
```

```python
import os
import numpy as np
from sklearn.utils.class_weight import compute_class_weight

train_path = "/content/Waste_Dataset_Split/train"

classes = [
    "battery",
    "biological",
    "brown-glass",
    "cardboard",
    "clothes",
    "green-glass",
    "metal",
    "paper",
    "plastic",
    "shoes",
    "trash",
    "white-glass"
]

# Get training labels
y_train = []

for index, cls in enumerate(classes):
    folder = os.path.join(train_path, cls)

    count = len([
        f for f in os.listdir(folder)
        if f.lower().endswith(
            (".jpg", ".jpeg", ".png", ".bmp", ".webp")
        )
    ])

    y_train.extend([index] * count)

# Calculate class weights
weights = compute_class_weight(
    class_weight="balanced",
    classes=np.arange(len(classes)),
    y=np.array(y_train)
)

class_weights = {
    i: float(weights[i])
    for i in range(len(classes))
}

print("CLASS WEIGHTS")
print("=" * 45)

for i, cls in enumerate(classes):
    print(f"{cls:15s} : {class_weights[i]:.4f}")
```

```python
import numpy as np
from PIL import Image

def predict_image(image, model, class_names):

    # Convert to RGB
    image = image.convert("RGB")

    # Resize exactly like training
    image = image.resize((224, 224))

    # Convert to numpy
    img_array = np.array(image, dtype=np.float32)

    # Normalize
    img_array = img_array / 255.0

    # Add batch dimension
    img_array = np.expand_dims(img_array, axis=0)

    # Prediction
    predictions = model.predict(img_array, verbose=0)[0]

    # Best prediction
    predicted_index = int(np.argmax(predictions))
    confidence = float(predictions[predicted_index])

    predicted_class = class_names[predicted_index]

    # Top 3
    top_indices = np.argsort(predictions)[::-1][:3]

    top_predictions = [
        (class_names[i], float(predictions[i]))
        for i in top_indices
    ]

    return predicted_class, confidence, top_predictions
```

```python
import tensorflow as tf
import json
import numpy as np
from PIL import Image
import os

# ==========================================
# LOAD MODEL
# ==========================================

model = tf.keras.models.load_model("/content/cnn_model.keras")

# ==========================================
# LOAD CLASS MAPPING
# ==========================================

with open("/content/class_names.json", "r") as f:
    class_data = json.load(f)

# Convert dictionary to correct index order
if isinstance(class_data, dict):
    class_names = [
        name for name, index in sorted(
            class_data.items(),
            key=lambda x: x[1]
        )
    ]
else:
    class_names = class_data

print("Classes:", class_names)

# ==========================================
# FIND TEST IMAGE
# ==========================================

test_dir = "/content/Waste_Dataset_Split/test"

found_image = None
actual_class = None

for cls in class_names:

    folder = os.path.join(test_dir, cls)

    if os.path.exists(folder):

        files = [
            f for f in os.listdir(folder)
            if f.lower().endswith(
                (".jpg", ".jpeg", ".png", ".webp")
            )
        ]

        if files:
            found_image = os.path.join(folder, files[0])
            actual_class = cls
            break

print("\nActual class:", actual_class)
print("Image:", found_image)

# ==========================================
# PREPROCESS IMAGE
# ==========================================

image = Image.open(found_image).convert("RGB")
image = image.resize((224, 224))

img = np.array(
    image,
    dtype=np.float32
) / 255.0

img = np.expand_dims(img, axis=0)

# ==========================================
# PREDICTION
# ==========================================

prediction = model.predict(
    img,
    verbose=0
)[0]

# ==========================================
# TOP 3
# ==========================================

top3 = np.argsort(prediction)[::-1][:3]

print("\nPREDICTION")
print("=" * 40)

for i in top3:

    i = int(i)

    print(
        f"{class_names[i]:15s} : "
        f"{prediction[i] * 100:.2f}%"
    )

print("\n" + "=" * 40)
print("Actual     :", actual_class)
print("Predicted  :", class_names[int(np.argmax(prediction))])
print(
    "Confidence :",
    f"{prediction[np.argmax(prediction)] * 100:.2f}%"
)
```

```python
import tensorflow as tf
import numpy as np
import json
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report

# ============================================================
# 1. LOAD MODEL
# ============================================================

model = tf.keras.models.load_model("/content/cnn_model.keras")

# ============================================================
# 2. CLASS NAMES
# ============================================================

with open("/content/class_names.json", "r") as f:
    class_data = json.load(f)

if isinstance(class_data, dict):
    class_names = [
        name
        for name, index in sorted(
            class_data.items(),
            key=lambda x: x[1]
        )
    ]
else:
    class_names = class_data

print("Classes:", class_names)

# ============================================================
# 3. TEST DATA
# ============================================================

test_dir = "/content/Waste_Dataset_Split/test"

test_datagen = tf.keras.preprocessing.image.ImageDataGenerator(
    rescale=1.0 / 255.0
)

test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode="categorical",
    shuffle=False
)

# ============================================================
# 4. PREDICTIONS
# ============================================================

print("\nGenerating predictions...")

predictions = model.predict(
    test_generator,
    verbose=1
)

y_pred = np.argmax(predictions, axis=1)
y_true = test_generator.classes

# ============================================================
# 5. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_true,
    y_pred
)

# ============================================================
# 6. PLOT
# ============================================================

plt.figure(figsize=(13, 11))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title(
    "Confusion Matrix - Waste Classification CNN",
    fontsize=16
)

plt.xlabel("Predicted Class", fontsize=12)
plt.ylabel("Actual Class", fontsize=12)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.yticks(
    rotation=0
)

plt.tight_layout()

plt.show()

# ============================================================
# 7. CLASSIFICATION REPORT
# ============================================================

print("\nCLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_true,
        y_pred,
        target_names=class_names,
        digits=2
    )
)
```
